# Example using CMEMS scatterometry data for Typhoon Lingling 2019

## Scatterometer Wind
- Looking through the CMEMS Marine Data Store to find dataset: https://data.marine.copernicus.eu/products?q=wind_speed&pk_vid=bfd50509683fc0421764148768aca348

```python
repo = CMEMSSatObsRepository(dataset_id="cmems_obs-wind_glo_phy_my_l3-metopa-ascat-des-0.125deg_P1D-i")

start_time="2019-09-01"
end_time="2019-09-30"

lon_min = 105
lat_min = -5
lon_max = 145
lat_max = 40

area = f"bbox={lon_min},{lat_min},{lon_max},{lat_max}"

data = repo.get_satobs_data(
    start_time=start_time,
    end_time=end_time,
    area=area,
)

df = data.df
wind_df = df[(df.index > pd.to_datetime('2019-09-04 01:22:00')) & (df.index < pd.to_datetime('2019-09-04 01:29:00')) & (df.longitude > 115)]
wind_df.to_csv(r'..\tests\data\UV_Lingling_2019.csv')
```

```python
repo = CMEMSSatObsRepository(dataset_id="cnes_obs-wave_glo_phy-spc_nrt_cfo-l3_PT10S")

data = repo.get_satobs_data(
    start_time=start_time,
    end_time=end_time,
    area=area,
)

df = data.df
spec_df = df[(df.index > pd.to_datetime('2019-09-04 00:00:00')) & (df.index < pd.to_datetime('2019-09-04 23:59:00'))]
spec_df.to_csv(r'..\tests\data\Spec_Lingling_2019.csv')
```

```python
repo = CMEMSSatObsRepository(dataset_id="cmems_obs-wave_glo_phy-swh_my_cfo-l3_PT1S")

data = repo.get_satobs_data(
    start_time=start_time,
    end_time=end_time,
    area=area,
)

df = data.df
wave_df = df[(df.index > pd.to_datetime('2019-09-04 00:00:00')) & (df.index < pd.to_datetime('2019-09-04 23:59:00'))]
wave_df.to_csv(r'..\tests\data\Hm0_Lingling_2019.csv')
```
